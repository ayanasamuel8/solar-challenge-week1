```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore

# Load the dataset
try:
    df_benin = pd.read_csv('../data/benin-malanville.csv') # Assuming data is in a parent data folder
    df_benin['Timestamp'] = pd.to_datetime(df_benin['Timestamp'])
    df_benin.set_index('Timestamp', inplace=True)
except FileNotFoundError:
    print("Ensure the dataset is located in the specified directory.")

print("Data loaded successfully. Shape:", df_benin.shape)
df_benin.head()
```

```python
try:
    df_sieraleone = pd.read_csv('../data/sierraleone-bumbuna.csv') # Assuming data is in a parent data folder
    df_sieraleone['Timestamp'] = pd.to_datetime(df_sieraleone['Timestamp'])
    df_sieraleone.set_index('Timestamp', inplace=True)
except FileNotFoundError:
    print("Ensure the dataset is located in the specified directory.")

print("Data loaded successfully. Shape:", df_sieraleone.shape)
df_sieraleone.head()
```

```python
try:
    df_togo = pd.read_csv('../data/togo-dapaong_qc.csv') # Assuming data is in a parent data folder
    df_togo['Timestamp'] = pd.to_datetime(df_togo['Timestamp'])
    df_togo.set_index('Timestamp', inplace=True)
except FileNotFoundError:
    print("Ensure the dataset is located in the specified directory.")

print("Data loaded successfully. Shape:", df_togo.shape)
df_togo.head()
```

```python
# Get a statistical overview of the key performance indicators
print("Financial & Operational Metrics Overview:")
print(df_benin[['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'Tamb', 'WS']].describe())

# Identify missing data, which represents gaps in our operational knowledge
print("\nMissing Data Report (%):")
missing_percentage = (df_benin.isna().sum() / len(df_benin)) * 100
print(missing_percentage[missing_percentage > 5])
```

```python
# Get a statistical overview of the key performance indicators
print("Financial & Operational Metrics Overview:")
print(df_sieraleone[['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'Tamb', 'WS']].describe())

# Identify missing data, which represents gaps in our operational knowledge
print("\nMissing Data Report (%):")
missing_percentage = (df_sieraleone.isna().sum() / len(df_sieraleone)) * 100
print(missing_percentage[missing_percentage > 5])
```

```python
# Get a statistical overview of the key performance indicators
print("Financial & Operational Metrics Overview:")
print(df_togo[['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'Tamb', 'WS']].describe())

# Identify missing data, which represents gaps in our operational knowledge
print("\nMissing Data Report (%):")
missing_percentage = (df_togo.isna().sum() / len(df_togo)) * 100
print(missing_percentage[missing_percentage > 5])
```

```python
# Calculate Z-scores for key performance and risk indicators
cols_to_check = ['GHI', 'DNI', 'DHI', 'ModA', 'ModB', 'WS', 'WSgust']
for col in cols_to_check:
    df_benin[f'{col}_zscore'] = np.abs(zscore(df_benin[col].dropna()))

# Flag rows with outliers for potential deeper investigation
outlier_rows = df_benin[df_benin.filter(like='_zscore').gt(3).any(axis=1)]
print(f"Identified {len(outlier_rows)} potential outlier data points (|Z|>3).")

# Basic Cleaning Strategy: Impute critical missing values with median
# For a financial model, a more sophisticated imputation might be needed, but median is a robust start.
for col in ['GHI', 'DNI', 'DHI', 'Tamb']:
    if df_benin[col].isna().any():
        median_val = df_benin[col].median()
        df_benin[col].fillna(median_val, inplace=True)

# Export the cleaned data for future modeling and comparison
# This cleaned dataset is the "single source of truth" for our financial analysis.
df_benin_cleaned = df_benin.drop(columns=df_benin.filter(like='_zscore').columns)
# df_benin_cleaned.to_csv(f'../data/[country_name]_clean.csv') # Uncomment when ready
print("\nData cleaned and prepared for analysis.")
```
